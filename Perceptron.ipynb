# Ejercicio 1.2

Supongamos que cada correo electrónico se representa mediante un vector de características donde cada componente corresponde a la frecuencia de aparición de una palabra clave. El perceptrón clasifica un correo como spam ($+1$) o no spam ($-1$) mediante

$$
h(\mathbf{x})=\operatorname{sign}(\mathbf{w}^T\mathbf{x}).
$$

Cada peso $w_i$ mide la influencia de la palabra correspondiente sobre la decisión del clasificador.

---

## (a) ¿Qué palabras tendrán un peso positivo grande?

Las palabras que aparecen con mucha frecuencia en mensajes de **spam** recibirán un **peso positivo grande**, ya que aumentan la probabilidad de que el producto interno

$$
\mathbf{w}^T\mathbf{x}
$$

sea positivo.

Ejemplos típicos son:

- free
- win
- winner
- prize
- lottery
- congratulations
- click
- offer
- limited
- urgent
- cash
- discount
- bonus
- crypto
- investment

Estas palabras son altamente indicativas de mensajes publicitarios o fraudulentos. Cuando aparecen varias de ellas en un correo, su contribución al producto interno incrementa significativamente la puntuación del perceptrón, favoreciendo la clasificación como spam.

---

## (b) ¿Qué palabras tendrán un peso negativo?

Las palabras asociadas con correos legítimos tendrán **pesos negativos**, ya que disminuyen el valor del producto interno y favorecen la clasificación como correo no deseado.

Algunos ejemplos son:

- meeting
- project
- report
- class
- homework
- university
- schedule
- invoice
- research
- attached
- professor
- document
- thanks
- regards

Estas palabras aparecen con frecuencia en comunicaciones laborales o académicas y rara vez en correos de spam.

---

## (c) ¿Qué parámetro del perceptrón afecta directamente cuántos mensajes cercanos a la frontera son clasificados como spam?

El parámetro que controla esta decisión es el **bias** o **término de sesgo** $w_0$.

La función de decisión del perceptrón es

$$
h(\mathbf{x})=
\operatorname{sign}
\left(
w_0+\sum_{i=1}^{d}w_i x_i
\right).
$$

El bias desplaza el hiperplano de decisión sin modificar su orientación.

- Si $w_0$ aumenta, resulta más difícil que un mensaje sea clasificado como spam.
- Si $w_0$ disminuye, más mensajes serán clasificados como spam.

Por lo tanto, el bias determina el umbral de decisión y controla cuántos ejemplos cercanos a la frontera quedan en una u otra clase.

# Ejercicio 1.3

La regla de actualización del perceptrón es

$$
\boxed{
\mathbf{w}(t+1)
=
\mathbf{w}(t)
+
y(t)\mathbf{x}(t)
}
$$

donde

- $y(t)\in\{-1,+1\}$,
- $\mathbf{x}(t)$ es un ejemplo mal clasificado.

---

## (a) Demuestre que

$$
y(t)\mathbf{w}(t)^T\mathbf{x}(t)<0.
$$

### Demostración

El ejemplo $\mathbf{x}(t)$ está mal clasificado.

La predicción del perceptrón es

$$
\hat y
=
\operatorname{sign}
\left(
\mathbf{w}(t)^T\mathbf{x}(t)
\right).
$$

Como la clasificación es incorrecta,

$$
\hat y
\neq
y(t).
$$

Existen únicamente dos posibilidades.

### Caso 1

Si

$$
y(t)=+1,
$$

el perceptrón predice

$$
-1,
$$

por lo que

$$
\mathbf{w}(t)^T\mathbf{x}(t)<0.
$$

Entonces

$$
y(t)\mathbf{w}(t)^T\mathbf{x}(t)
=
(+1)
\mathbf{w}(t)^T\mathbf{x}(t)
<
0.
$$

### Caso 2

Si

$$
y(t)=-1,
$$

el perceptrón predice

$$
+1,
$$

de modo que

$$
\mathbf{w}(t)^T\mathbf{x}(t)>0.
$$

Multiplicando por $y(t)$,

$$
(-1)
\mathbf{w}(t)^T\mathbf{x}(t)
<
0.
$$

En ambos casos,

$$
\boxed{
y(t)\mathbf{w}(t)^T\mathbf{x}(t)<0.
}
$$

---

## (b) Demuestre que

$$
y(t)\mathbf{w}(t+1)^T\mathbf{x}(t)
>
y(t)\mathbf{w}(t)^T\mathbf{x}(t).
$$

### Demostración

Usando la regla de actualización,

$$
\mathbf{w}(t+1)
=
\mathbf{w}(t)
+
y(t)\mathbf{x}(t).
$$

Multiplicando por $\mathbf{x}(t)$,

$$
\mathbf{w}(t+1)^T\mathbf{x}(t)
=
\mathbf{w}(t)^T\mathbf{x}(t)
+
y(t)\|\mathbf{x}(t)\|^2.
$$

Multiplicando toda la expresión por $y(t)$,

$$
y(t)\mathbf{w}(t+1)^T\mathbf{x}(t)
=
y(t)\mathbf{w}(t)^T\mathbf{x}(t)
+
y(t)^2\|\mathbf{x}(t)\|^2.
$$

Como

$$
y(t)^2=1,
$$

obtenemos

$$
y(t)\mathbf{w}(t+1)^T\mathbf{x}(t)
=
y(t)\mathbf{w}(t)^T\mathbf{x}(t)
+
\|\mathbf{x}(t)\|^2.
$$

Dado que

$$
\|\mathbf{x}(t)\|^2>0,
$$

se concluye que

$$
\boxed{
y(t)\mathbf{w}(t+1)^T\mathbf{x}(t)
>
y(t)\mathbf{w}(t)^T\mathbf{x}(t).
}
$$

---

## (c) Argumente que el paso de

$$
\mathbf{w}(t)
\rightarrow
\mathbf{w}(t+1)
$$

es un movimiento en la dirección correcta.

El apartado (b) demuestra que, después de una actualización,

$$
y(t)\mathbf{w}(t+1)^T\mathbf{x}(t)
>
y(t)\mathbf{w}(t)^T\mathbf{x}(t).
$$

Esto significa que el valor del margen del ejemplo aumenta.

- Si el ejemplo era positivo, el producto interno se vuelve más positivo.
- Si el ejemplo era negativo, el producto interno se vuelve más negativo.

En ambos casos, el hiperplano se desplaza en la dirección que favorece la clasificación correcta del ejemplo.

Geométricamente, la actualización mueve el hiperplano de decisión hacia la región donde se encuentra el punto mal clasificado, reduciendo el error cometido en esa observación.

Repetir este procedimiento para todos los ejemplos mal clasificados hace que el hiperplano se acerque progresivamente a uno que separa correctamente las dos clases, siempre que los datos sean linealmente separables.

Por ello, la actualización del perceptrón puede interpretarse como un **movimiento en la dirección correcta**, ya que incrementa la probabilidad de clasificar correctamente el ejemplo que originó la actualización y constituye la base del Teorema de Convergencia del Perceptrón.

```python
import numpy as np
import matplotlib.pyplot as plt

# =====================================================
# Generación de datos
# =====================================================

np.random.seed(42)

N = 100

# Puntos aleatorios en [-1,1]x[-1,1]
X = np.random.uniform(-1, 1, (N, 2))

# Agregar término de bias
X_bias = np.hstack((np.ones((N,1)), X))

# Recta objetivo (target function)
w_target = np.array([-0.2, 0.6, -0.8])

# Etiquetas verdaderas
y = np.sign(X_bias @ w_target)
y[y == 0] = 1

# =====================================================
# Algoritmo del Perceptrón
# =====================================================

def perceptron(X, y, max_iterations=10000):

    n_samples, n_features = X.shape

    # Inicialización aleatoria
    w = np.random.randn(n_features)

    iterations = 0
    errors_history = []

    while iterations < max_iterations:

        # Predicción
        predictions = np.sign(X @ w)
        predictions[predictions == 0] = 1

        # Índices mal clasificados
        misclassified = np.where(predictions != y)[0]

        errors_history.append(len(misclassified))

        # Convergencia
        if len(misclassified) == 0:
            break

        # Elegir un punto aleatorio mal clasificado
        idx = np.random.choice(misclassified)

        # Regla de actualización
        w = w + y[idx] * X[idx]

        iterations += 1

    return w, iterations, errors_history


w_final, iterations, errors = perceptron(X_bias, y)

print("Vector objetivo:")
print(w_target)

print("\nVector aprendido:")
print(w_final)

print(f"\nNúmero de iteraciones: {iterations}")
```

```python
def plot_result(X, y, w_target, w_learned):

    plt.figure(figsize=(8,8))

    plt.scatter(X[y==1,1], X[y==1,2],
                color="red", label="Clase +1")

    plt.scatter(X[y==-1,1], X[y==-1,2],
                color="blue", label="Clase -1")

    x = np.linspace(-1,1,100)

    # Recta verdadera
    yt = -(w_target[0] + w_target[1]*x)/w_target[2]

    # Recta aprendida
    yl = -(w_learned[0] + w_learned[1]*x)/w_learned[2]

    plt.plot(x, yt, 'k--', label="Recta objetivo")
    plt.plot(x, yl, 'g', linewidth=2, label="Perceptrón")

    plt.xlim(-1,1)
    plt.ylim(-1,1)
    plt.legend()
    plt.grid(True)
    plt.show()

plot_result(X_bias, y, w_target, w_final)
```

```python
plt.figure(figsize=(7,4))

plt.plot(errors, linewidth=2)

plt.xlabel("Iteración")
plt.ylabel("Número de errores")
plt.title("Convergencia del Perceptrón")

plt.grid(True)

plt.show()
```

# Explicación del Algoritmo de Aprendizaje del Perceptrón

## Modelo del Perceptrón

Sea un conjunto de datos de entrenamiento formado por pares

$$
(\mathbf{x}_i,y_i), \qquad i=1,\ldots,N,
$$

donde:

- $\mathbf{x}_i \in \mathbb{R}^{d+1}$ representa un vector de entrada (incluyendo el término de sesgo o *bias*).
- $y_i \in \{-1,+1\}$ es la etiqueta de clase.

El clasificador lineal está definido por

$$
h(\mathbf{x}) = \operatorname{sign}(\mathbf{w}^T \mathbf{x}),
$$

donde

$$
\mathbf{w} = (w_0,w_1,\ldots,w_d)^T
$$

es el vector de pesos que debe aprender el algoritmo.

---

## Regla de actualización

Cuando un punto es clasificado incorrectamente, el perceptrón actualiza sus pesos mediante

$$
\mathbf{w}
\leftarrow
\mathbf{w}
+
y_i \mathbf{x}_i.
$$

Esta expresión resume los dos casos del algoritmo:

- Si un ejemplo positivo ($y=+1$) fue clasificado como negativo,

$$
\mathbf{w}
\leftarrow
\mathbf{w}
+
\mathbf{x}.
$$

- Si un ejemplo negativo ($y=-1$) fue clasificado como positivo,

$$
\mathbf{w}
\leftarrow
\mathbf{w}
-
\mathbf{x}.
$$

Cada actualización modifica el hiperplano de decisión con el objetivo de corregir el error cometido.

---

# Preguntas sobre la convergencia del Perceptrón

## 1. ¿El algoritmo encuentra los parámetros correctos?

La respuesta es **sí**, siempre que los datos sean **linealmente separables**.

Si existe un hiperplano

$$
\mathbf{w}^*
$$

capaz de separar perfectamente las dos clases, el perceptrón garantiza encontrar algún vector de pesos

$$
\mathbf{w}
$$

que clasifique correctamente todos los ejemplos del conjunto de entrenamiento.

Sin embargo, el algoritmo **no garantiza encontrar exactamente el mismo hiperplano** $\mathbf{w}^*$.

En general existen infinitas soluciones válidas, por lo que el vector obtenido dependerá de:

- la inicialización de los pesos,
- el orden en que se seleccionan los ejemplos,
- las actualizaciones realizadas durante el entrenamiento.

En consecuencia, el perceptrón encuentra un clasificador correcto, aunque no necesariamente el óptimo.

---

## 2. ¿El algoritmo termina en un número finito de pasos?

Sí.

El **Teorema de Convergencia del Perceptrón**, propuesto por Frank Rosenblatt (1958), establece que si los datos son linealmente separables, el algoritmo siempre converge después de un número finito de actualizaciones.

Sea:

- $R$ una cota superior para la norma de todos los ejemplos,

$$
\|\mathbf{x}_i\| \le R,
$$

- $\gamma > 0$ el margen de separación entre las clases.

Entonces el número máximo de actualizaciones satisface

$$
T
\le
\left(\frac{R}{\gamma}\right)^2.
$$

Por tanto, el algoritmo necesariamente termina.

Si los datos **no son linealmente separables**, el algoritmo puede seguir actualizando los pesos indefinidamente y no existe garantía de convergencia.

---

## 3. ¿Por qué funciona?

La demostración del Teorema de Convergencia se basa en dos propiedades fundamentales.

### Paso 1. Los pesos se acercan al hiperplano correcto

Sea

$$
\mathbf{w}^*
$$

un hiperplano que separa correctamente todos los datos.

Cada vez que el algoritmo comete un error, actualiza

$$
\mathbf{w}_{t+1}
=
\mathbf{w}_t
+
y_i\mathbf{x}_i.
$$

Tomando el producto interno con $\mathbf{w}^*$,

$$
\mathbf{w}_{t+1}^T\mathbf{w}^*
=
\mathbf{w}_t^T\mathbf{w}^*
+
y_i\mathbf{x}_i^T\mathbf{w}^*.
$$

Como todos los ejemplos están correctamente separados,

$$
y_i\mathbf{x}_i^T\mathbf{w}^*
\ge
\gamma,
$$

donde $\gamma$ es el margen de separación.

Por tanto,

$$
\mathbf{w}_{t+1}^T\mathbf{w}^*
\ge
\mathbf{w}_t^T\mathbf{w}^*
+
\gamma.
$$

Después de $T$ iteraciones,

$$
\mathbf{w}_T^T\mathbf{w}^*
\ge
T\gamma.
$$

Esto significa que el vector de pesos se orienta progresivamente hacia el hiperplano correcto.

---

### Paso 2. La norma de los pesos crece lentamente

El cuadrado de la norma del vector actualizado es

$$
\|\mathbf{w}_{t+1}\|^2
=
\|\mathbf{w}_t\|^2
+
2y_i\mathbf{w}_t^T\mathbf{x}_i
+
\|\mathbf{x}_i\|^2.
$$

Como el punto estaba mal clasificado,

$$
y_i\mathbf{w}_t^T\mathbf{x}_i
\le
0,
$$

entonces

$$
\|\mathbf{w}_{t+1}\|^2
\le
\|\mathbf{w}_t\|^2
+
R^2.
$$

Después de $T$ actualizaciones,

$$
\|\mathbf{w}_T\|
\le
R\sqrt{T}.
$$

Es decir, la magnitud del vector de pesos aumenta de forma controlada.

---

### Paso 3. Aplicación de la desigualdad de Cauchy-Schwarz

La desigualdad de Cauchy-Schwarz establece que

$$
\mathbf{w}_T^T\mathbf{w}^*
\le
\|\mathbf{w}_T\|
\,
\|\mathbf{w}^*\|.
$$

Sin pérdida de generalidad se puede asumir

$$
\|\mathbf{w}^*\| = 1.
$$

Combinando los dos resultados anteriores,

$$
T\gamma
\le
R\sqrt{T}.
$$

Elevando ambos lados al cuadrado,

$$
T
\le
\left(\frac{R}{\gamma}\right)^2.
$$

Como esta cantidad es finita, el número de errores también lo es.

En consecuencia, el algoritmo necesariamente converge cuando los datos son linealmente separables.

---

# Conclusiones

- El perceptrón aprende un hiperplano que separa correctamente los datos siempre que exista una separación lineal.
- El algoritmo converge en un número finito de iteraciones si los datos son linealmente separables.
- La convergencia se demuestra mostrando que el vector de pesos se acerca continuamente al hiperplano correcto mientras su norma permanece acotada, lo que permite obtener una cota superior para el número de actualizaciones.
